## 5. Methods

### 5.1 Preprocessing and filtering

* Rare taxa are filtered to reduce noise: taxa must be present in **≥3 samples** (your run retained **298 taxa**, which is a reasonable core size for scanning).
* Outcome (Observed), (Y): total abundance of each taxon in the degraded condition.
* Expectation, (E): expected abundance of each taxon based on the reference condition, **scaled by sequencing depth**:
  [
  E_i = (\text{Reference}_i \times \text{depth_scaling}) + \epsilon
  ]
  Your run estimated a depth scaling factor of **~1.3037**, used to align expected counts to target sequencing depth.

### 5.2 Feature space for scanning

Taxa are described using categorical coordinates:

* phylum
* class
* order
* family
  Missing values are coded as “Unclassified” so they remain eligible for selection.

### 5.3 MDSS scanning (Poisson)

* Scanner: MDSS with Poisson scoring using ((Y, E))
* Two scan directions:

  * **Negative direction**: identifies taxa subsets with (Y \ll E) (depletion; “lost recruit” guild)
  * **Positive direction**: identifies taxa subsets with (Y \gg E) (enrichment; “invader” guild)
* Penalty grid search: penalty values from low to high to balance subset size vs score.
  This yields a set of guilds across penalties and supports identification of a **consensus core** via intersection across penalty-stable solutions.

### 5.4 Reporting metrics (guild-level effect sizes)

For each guild:

* **LLR score** (MDSS objective)
* **Guild size**
* **Observed** and **Expected** totals
* **Relative Risk (RR)**:
  [
  RR = \frac{\sum Y}{\sum E}
  ]
  Interpretation:

  * RR < 1 = depletion (lost recruit guild)
  * RR > 1 = enrichment (invader guild)

**Important correction for rigor:**
For the enriched scan, computing “% of total loss” using ((E - Y)) will produce negative values because enriched taxa have (Y > E). This is not interpretable as “loss contribution.”
Instead, report:

* For depleted guilds: “% of total depletion explained” using (\sum(E-Y)) where (E>Y)
* For enriched guilds: “% of total enrichment explained” using (\sum(Y-E)) where (Y>E)

This correction aligns the metric with scan direction.

### 5.5 Database-inferred pathway analysis (MetaCyc-style)

Using the consensus core genera:

* Map genera to curated pathway identifiers (e.g., nitrogen fixation, ammonia oxidation, stress response, cell wall synthesis).
* Compute pathway abundance by summing genus abundance contributions.
* Report pathway shifts using:

  * log2 fold change (degraded/reference)
  * total predicted abundance (for biological relevance)

Two visualization outputs (as in your figures):

1. **Differential abundance dot plot**: log2FC vs total abundance
2. **Pathway saturation heatmap**: relative abundance by condition (row-wise normalization)

Because this is inference from taxonomy-to-function mapping, results will be framed as **predicted functional capacity**, not measured function.

```python
def parse_faprotax_db(db_txt_path):
    """
    Parses a FAPROTAX-like database text file into a long table:
    columns = ['Function','Taxon','Rank','Genus']

    Assumptions (common in FAPROTAX db files):
    - File is organized in blocks per function
    - Taxa are listed beneath a function name
    - Taxa can appear at multiple ranks; we extract Genus if possible

    If your specific db format differs, we can adjust the regex rules.
    """
    functions = []
    with open(db_txt_path, "r", encoding="utf-8", errors="ignore") as f:
        lines = [ln.strip() for ln in f.readlines()]

    current_func = None
    for ln in lines:
        if not ln or ln.startswith("#"):
            continue

        # Heuristic: function headers often have no semicolons and no taxonomy prefixes
        # and are not obviously taxonomic strings.
        # Many FAPROTAX db files separate blocks with a function name line.
        if (";" not in ln) and (ln.lower() == ln) and (" " not in ln) and (len(ln) < 80):
            # e.g. "nitrification", "nitrogen_fixation", "chemoheterotrophy"
            current_func = ln
            continue

        if current_func is None:
            continue

        # Taxon line: often like "Genus" or "Genus species" or "Bacteria; ...; Genus"
        taxon = ln

        # Extract genus: last token after semicolons if present
        if ";" in taxon:
            genus_guess = taxon.split(";")[-1].strip()
        else:
            genus_guess = taxon.split()[0].strip()

        # Clean genus_guess
        genus_guess = genus_guess.replace("[","").replace("]","")
        genus_guess = genus_guess.replace("unclassified","unclassified").strip()

        functions.append({
            "Function": current_func,
            "Taxon": taxon,
            "Genus": genus_guess
        })

    df_map = pd.DataFrame(functions).dropna()
    # Keep only plausible genus tokens (simple screen)
    df_map = df_map[df_map["Genus"].str.len() > 1]
    return df_map[["Function","Genus"]].drop_duplicates()
```

## 5A) Function overrepresentation (Fisher’s exact test)

```python
def function_overrepresentation(candidate_genera, background_genera, fap_map, min_support=3):
    """
    Enrichment of functions among candidate genera vs background genera.

    min_support: require at least this many mapped genera in candidate+background for stability.
    """
    # Map genus -> set(functions)
    g2f = fap_map.groupby("Genus")["Function"].apply(set).to_dict()

    # Only consider genera that are mapped in the db
    cand = set([g for g in candidate_genera if g in g2f])
    back = set([g for g in background_genera if g in g2f])

    # Universe for test = mapped background
    universe = back.copy()

    # Function list across universe
    all_funcs = sorted(set(fap_map.loc[fap_map["Genus"].isin(universe), "Function"]))

    rows = []
    for func in all_funcs:
        # a = cand with func
        a = sum(func in g2f[g] for g in cand)
        # b = cand without func
        b = len(cand) - a
        # c = non-cand with func
        noncand = universe - cand
        c = sum(func in g2f[g] for g in noncand)
        # d = non-cand without func
        d = len(noncand) - c

        if (a + c) < min_support:
            continue

        odds, p = fisher_exact([[a,b],[c,d]], alternative="greater")
        rows.append({
            "Function": func,
            "Cand_with_func": a,
            "Cand_total_mapped": len(cand),
            "Back_with_func": a + c,
            "Back_total_mapped": len(universe),
            "OddsRatio": odds,
            "P": p
        })

    out = pd.DataFrame(rows)
    if out.empty:
        return out

    out["FDR"] = multipletests(out["P"], method="fdr_bh")[1]
    out = out.sort_values(["FDR","P","OddsRatio"], ascending=[True,True,False])
    return out
```

```python
def clean_genus_token(g):
    g = g.strip()
    g = g.replace("[","").replace("]","")
    return g

def extract_genus_from_pattern(pattern):
    """
    pattern examples:
      *Proteobacteria*Nitrosomonas*
      *Bacillus*pumilus*
      *Azospirillum*
    We take the last non-empty token between asterisks as genus/species.
    Then we keep just the first word as Genus.
    """
    tokens = [t for t in pattern.split("*") if t.strip()]
    if not tokens:
        return None
    last = clean_genus_token(tokens[-1])
    genus = last.split()[0].strip()
    return genus if genus else None

def parse_faprotax_genus_function(db_path):
    """
    Reads FAPROTAX edited DB and returns DataFrame with columns:
      Function, Genus

    In your snippet, taxon lines look like:
      *Bacillus*pumilus*   #manganese_oxidation
    So we parse lines containing '#' to get function reliably.
    """
    rows = []
    with open(db_path, "r", encoding="utf-8", errors="ignore") as f:
        for ln in f:
            ln = ln.strip()
            if not ln or ln.startswith("#"):
                continue

            # We focus on mapping lines that contain '#'
            if "#" in ln and ln.startswith("*"):
                left, right = ln.split("#", 1)
                pattern = left.strip()
                func = right.strip()
                # Some files include extra comments after function; keep first token
                func = func.split()[0].strip()

                genus = extract_genus_from_pattern(pattern)
                if genus:
                    rows.append({"Function": func, "Genus": genus})

    df = pd.DataFrame(rows).drop_duplicates()
    return df

def function_enrichment(candidate_genera, background_genera, fap_map, min_support=3):
    """
    Fisher exact enrichment for each function:
      in candidate vs in background (mapped universe).
    candidate_genera: set of genera
    background_genera: set of genera (universe)
    fap_map: DataFrame with Function, Genus
    """
    g2f = fap_map.groupby("Genus")["Function"].apply(set).to_dict()
    candidate = set([g for g in candidate_genera if g in g2f])
    universe  = set([g for g in background_genera if g in g2f])

    funcs = sorted(set(fap_map.loc[fap_map["Genus"].isin(universe), "Function"]))
    rows = []
    noncand = universe - candidate

    for func in funcs:
        a = sum(func in g2f[g] for g in candidate)     # cand with func
        b = len(candidate) - a                         # cand without func
        c = sum(func in g2f[g] for g in noncand)       # noncand with func
        d = len(noncand) - c                           # noncand without func

        if (a + c) < min_support:
            continue

        odds, p = fisher_exact([[a, b], [c, d]], alternative="greater")
        rows.append({
            "Function": func,
            "Cand_with_func": a,
            "Cand_total_mapped": len(candidate),
            "Bg_with_func": a + c,
            "Bg_total_mapped": len(universe),
            "OddsRatio": odds,
            "P": p
        })

    out = pd.DataFrame(rows)
    if out.empty:
        return out

    out["FDR"] = multipletests(out["P"].values, method="fdr_bh")[1]
    out = out.sort_values(["FDR","P","OddsRatio"], ascending=[True,True,False]).reset_index(drop=True)
    return out

def effect_size_cohens_d(x, y):
    x = np.asarray(x); y = np.asarray(y)
    if len(x) < 2 or len(y) < 2:
        return np.nan
    vx = x.var(ddof=1); vy = y.var(ddof=1)
    pooled = ((len(x)-1)*vx + (len(y)-1)*vy) / (len(x)+len(y)-2 + 1e-12)
    return (x.mean() - y.mean()) / np.sqrt(pooled + 1e-12)

    cohen = effect_size_cohens_d(d, h)  # degraded - healthy

def compute_func_stats_for_genus_set(genus_set, label):
    sub = df_counts_long[df_counts_long["genus"].isin(genus_set)].copy()
    sub = sub.merge(fap_map, left_on="genus", right_on="Genus", how="inner")
    fs = sub.groupby(["Sample","Condition","Function"])["RelAbundance"].sum().reset_index()
    mat = fs.pivot_table(index=["Sample","Condition"], columns="Function", values="RelAbundance", fill_value=0)

    rows = []
    for func in mat.columns:
        h = mat.xs("Healthy", level="Condition")[func].values
        d = mat.xs("Degraded", level="Condition")[func].values
        log2fc = np.log2((d.mean() + 1e-12) / (h.mean() + 1e-12))
        cohen = effect_size_cohens_d(d, h)
        try:
            p = mannwhitneyu(h, d, alternative="two-sided").pvalue
        except Exception:
            p = np.nan

        rows.append({
            "Set": label,
            "Function": func,
            "Mean_Healthy": h.mean(),
            "Mean_Degraded": d.mean(),
            "Log2FC_Degraded_vs_Healthy": log2fc,
            "Cohens_d_Degraded_vs_Healthy": cohen,
            "MWU_p": p,
            "TotalMean": h.mean() + d.mean()
        })

    out = pd.DataFrame(rows).sort_values("TotalMean", ascending=False)
    out["FDR"] = multipletests(out["MWU_p"].fillna(1.0), method="fdr_bh")[1]
    return out

def contribution_heatmap(genus_set, func_stats, title, topN_funcs=4, topN_genera=10, use_condition="Degraded"):
    # pick top enriched/depleted functions by abs(log2fc) among top mean
    d = func_stats.head(50).copy()
    d["abs_fc"] = d["Log2FC_Degraded_vs_Healthy"].abs()
    funcs = d.sort_values("abs_fc", ascending=False).head(topN_funcs)["Function"].tolist()

    sub = df_counts_long[df_counts_long["genus"].isin(genus_set)].copy()
    sub = sub.merge(fap_map, left_on="genus", right_on="Genus", how="inner")
    sub["Healthy_sum"] = sub[sub["Sample"].isin(healthy_use)].groupby("genus")["RelAbundance"].transform("sum")
    sub["Degraded_sum"] = sub[sub["Sample"].isin(degraded_use)].groupby("genus")["RelAbundance"].transform("sum")

    # Work with condition-specific totals by genus within function
    sub_cond = sub[sub["Condition"] == use_condition].copy()
    sub_cond = sub_cond[sub_cond["Function"].isin(funcs)]

    pivot = sub_cond.pivot_table(index="Function", columns="genus", values="RelAbundance", aggfunc="sum").fillna(0)
    # keep top genera overall
    top_cols = pivot.sum(axis=0).sort_values(ascending=False).head(topN_genera).index
    pivot = pivot[top_cols]

    # row-wise %
    pivot_pct = pivot.div(pivot.sum(axis=1).replace(0, np.nan), axis=0) * 100
    pivot_pct = pivot_pct.fillna(0)

    plt.figure(figsize=(12, 4.2))
    sns.heatmap(pivot_pct, cmap="viridis", annot=True, fmt=".0f", cbar_kws={"label": f"% contribution ({use_condition})"})
    plt.title(title)
    plt.ylabel("")
    plt.xlabel("Genus")
    plt.tight_layout()
    plt.show()

contribution_heatmap(lost_set, func_stats_lost, "Genus contributions to key inferred functions (Lost core)", topN_funcs=4)
contribution_heatmap(inv_set,  func_stats_inv,  "Genus contributions to key inferred functions (Invader core)", topN_funcs=4)
```

## 🧮 Three Ways to Demonstrate It

### ✅ Level 1 (Simplest – Already Strong Enough)

**Taxonomic enrichment test**

Steps:

1. Count number of genera per phylum in:

   * Invader Core
   * Lost Core
   * All genera (background)
2. Perform hypergeometric or Fisher’s exact test:

   * Is Actinobacteria overrepresented in Invader Core?

If significantly enriched → evidence of clustering.

This is easy and publishable.

---

### ✅ Level 2 (Better – Quantitative Clustering)

Construct a phylogenetic tree (even genus-level reference tree).

Calculate:

* Mean pairwise phylogenetic distance (MPD) among core taxa.
* Compare to MPD of randomly sampled sets of equal size.
* Use permutation (1000 iterations).

If observed MPD is significantly lower than random → phylogenetic clustering.

This is classical community phylogenetics.

---

### ✅ Level 3 (Advanced but Not Necessary)

Compute:

* Net Relatedness Index (NRI)
* Nearest Taxon Index (NTI)

Positive NRI → clustering.
Negative NRI → overdispersion.

For your paper, Level 1 or 2 is sufficient.

---

## 📊 How You Present It

Figure option:

* Bar plot showing phylum enrichment.
* OR histogram of null MPD distribution with observed line.

In Results, you would write:

> The Invader Core exhibited significant phylogenetic clustering (MPD_obs < MPD_null, p < 0.01), with overrepresentation of Actinobacteria (Fisher’s exact test, p < 0.001), indicating non-random taxonomic assembly under intensification.
