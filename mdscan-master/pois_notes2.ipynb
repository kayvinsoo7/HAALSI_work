## **3. Methods**

* **Field 2 (intensively cultivated field)**: long-term maize monoculture

From each field, **four rhizosphere samples** were collected from spatially distinct sites (north, south, east, west). In addition, **two bulk soil samples** were collected approximately 10 m from the cultivated area in each field. In total, **12 samples** were analyzed (6 rhizosphere, 6 bulk).

### **3.2 Soil Chemistry**

Soil physicochemical variables, including **pH and nitrate concentration**, were measured for all samples and used to contextualize microbial functional patterns. Consistent with prior reporting, the intensively cultivated field exhibited **lower pH and reduced nitrate availability**, indicative of soil degradation.

### **3.3 Microbial Community Data**

Microbial composition was assessed using **16S rRNA amplicon sequencing**. Taxa were filtered to retain those present in at least three samples to reduce sparsity and noise. Analyses were conducted at the **genus level**, reflecting the resolution of downstream functional inference.

### **3.4 Functional Trait Inference (FAPROTAX)**

Functional traits were inferred using the **FAPROTAX database**, which maps prokaryotic taxa to experimentally supported ecological functions. A cleaned and curated version of the database was used to avoid spurious labels (e.g., author names or citation artifacts). Trait abundances were calculated as **abundance-weighted sums** of mapped genera within each sample.

### **3.5 Statistical Analysis**

* **Group comparison**: Samples were grouped by field (reference vs intensively cultivated).
* **Effect sizes**: Log₂ fold changes (degraded vs reference) were calculated for each inferred trait.
* **Uncertainty estimation**: 95% bootstrap confidence intervals were generated to assess robustness.
* **Non-parametric testing**: Mann–Whitney U tests were used for exploratory inference, with false discovery rate (FDR) correction applied where appropriate.

All analyses emphasize **effect sizes and consistency** rather than reliance on nominal statistical significance, in line with best practices for microbiome studies with modest sample sizes.

## 1) Offset Poisson regression view (exactly what your E is)

In epidemiology, a common model is:

$
Y_i \sim \text{Poisson}(\mu_i)
$
$
\mu_i = \lambda \cdot \text{exposure}_i
$

Taking logs:


$\log(\mu_i) = \log(\lambda) + \log(\text{exposure}_i)$


That last term, $(\log(\text{exposure}_i))$, is the **offset** (fixed, known).

### Map to your case

* Outcome:
  
  
  $Y_i = \text{AG}_i = \sum \text{target\_cols}_i$
  
* Exposure (offset basis):
  
  $X_i = \text{GZ}_i = \sum \text{ref\_cols}_i$
  

### Your null model is: “same AG-per-GZ rate everywhere”

Define the **global rate**:

$\hat{\lambda} = \frac{\sum_i Y_i}{\sum_i X_i}$

Then the expected mean for each taxon is:


$E_i = \hat{\lambda} \cdot X_i$


That is *exactly* what your code does:

* `scaling_factor = depth_ag / depth_gz` is $(\hat{\lambda})$
* `E = GZ_i * scaling_factor` is $(E_i)$

### Important intuition

In the “everyone has same expected risk” disease scan, exposure is effectively constant: (X_i = 1).
Then:

$
E_i = \hat{\lambda} \cdot 1 = \hat{\lambda}
$

So all individuals share the same expected value.
Your implementation just uses **non-constant exposure** (GZ varies per taxon).

## 2) How MDSS is implicitly estimating a rate ratio (RR)

MDSS scans subsets (S) and compares **observed vs expected** inside the subset.

Let:

$
Y(S) = \sum_{i \in S} Y_i,\quad E(S) = \sum_{i \in S} E_i
$

### What RR means here

A natural “rate ratio” for a subset is:


$RR(S) = \frac{Y(S)}{E(S)}$


* If (RR(S) = 1): subset follows the null (AG proportional to GZ)
* If (RR(S) < 1): subset is **depleted** in AG relative to what GZ predicts
* If (RR(S) > 1): subset is **enriched**

That’s why your code prints:

```python
g_rr = g_obs / g_exp
```

### What MDSS is doing under the hood (conceptually)

For a candidate subset $(S)$, MDSS compares:

* **Null**: $(Y_i \sim \text{Poisson}(E_i))$ for everyone (RR = 1 everywhere)
* **Alternative**: in subset (S), the mean is multiplied by a factor (q)

$
Y_i \sim \text{Poisson}(q \cdot E_i) \quad \text{for } i \in S
$
$
Y_i \sim \text{Poisson}(E_i) \quad \text{for } i \notin S
$

Here (q) is the subset’s multiplicative effect — i.e. the subset’s **RR**.

The best-fitting (q) for a fixed subset (the MLE) is:

$
\hat{q}(S) = \frac{Y(S)}{E(S)}
$

So when you compute (g_rr = g_obs/g_exp), you’re computing the same thing MDSS is effectively fitting for that subset.

### One-liner that clicks

> MDSS is searching over subsets to find where a single multiplier (q) makes the Poisson model fit much better — and that (q) equals (Y(S)/E(S)), the subset RR.

## 3) How MDSS “gets each subset better” (what your library code is doing)

MDSS is doing **coordinate ascent** over features (`phylum`, `className`, `order`, `family`):

### Step 0: Start with a current subset rule

Example rule shape:

```python
subset = {"phylum": ["Verrucomicrobia"], "family": ["X", "Y"]}
```

### Step 1: Pick one feature to optimize (say `phylum`)

In `_scan_single_ascent`, it randomly chooses a feature not yet “flagged”.

### Step 2: Condition on the *other* features already in the subset

In `get_aggregates()`, it first filters the dataset to only rows matching the current subset *excluding the feature being optimized*.

So if your current subset already fixed `family`, and you are scanning `phylum`, it looks only at rows in those families, then asks:

> “Which phylum values should I include to maximize the score?”

### Step 3: For each value of that feature, compute (Observed sum, Expected sum)

Still in `get_aggregates()`:

For each phylum value (v), it computes over the rows with `phylum = v`:

* $(Y(v) = \sum Y_i)$
* $(E(v) = \sum E_i)$

This is the key: **subset observed and expected are sums of per-row values**.

### Step 4: Convert each feature value into a “candidate contribution interval”

For Poisson, the model inside a subset is:

$
Y_i \sim \text{Poisson}(qE_i) \quad \text{for } i\in S
$

So for a subset (S):

* $(Y(S) = \sum_{i\in S} Y_i)$
* $(E(S) = \sum_{i\in S} E_i)$
* the MLE:
  $
  \hat{q}(S)=\frac{Y(S)}{E(S)}
  $

And the score you pasted is:

$
\text{score}(S)=Y(S)\log(q) + \sum(E_i-qE_i) - \text{penalty}
$
Since (\sum(E_i-qE_i)=E(S)(1-q)), this is basically:
$
Y(S)\log(q) + E(S)(1-q) - \text{penalty}
$

### Step 5: ALTSS trick — choose the best set of values without brute forcing all combinations

This is what `choose_aggregates()` is doing:

* For each feature value (v), it computes a range ((q_{\min}(v), q_{\max}(v))) where that value would contribute **positively** to the score (after penalty).
* It collects all those thresholds.
* For each interval between thresholds, it selects all values whose ((q_{\min}, q_{\max})) covers that interval midpoint.

That’s why your subset can include **multiple** phyla/families at once, but still be found efficiently.

### Step 6: Update subset, repeat for another feature until no improvement

When a change increases score, flags reset and it continues until no single-feature update improves the score.

That is literally “getting each subset better”.
