* **Non-parametric testing**: Mann–Whitney U tests were used for exploratory inference, with false discovery rate (FDR) correction applied where appropriate.

All analyses emphasize **effect sizes and consistency** rather than reliance on nominal statistical significance, in line with best practices for microbiome studies with modest sample sizes.

### Limitations of Existing Analytical Approaches

Most standard microbiome analyses rely on differential abundance testing or multivariate ordination methods. These approaches face two major limitations in field-based soil studies. First, they often lack statistical power in modestly sized but carefully designed datasets, such as the present study with twelve spatially structured samples. Addressing this question requires methods that explicitly search for **groups of taxa that behave coherently across environmental conditions**, rather than testing taxa one by one.
